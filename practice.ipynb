```python
#password strength checking function
def check_password():
    x:str = input("Enter password: ")
    sp_char:str = ['@','#','$','%','&','^','!','~']
    if len(x) < 8:
        print("Password should be at least 8 characters long.")
    elif not any(char.isupper() for char in x):
        print("Password should contain at least one uppercase letter.")
    elif "passwords" in x.lower():
        print("Password should not contain 'passwords'.")
    elif not any(char.isdigit() for char in x):
        print("Password should contain at least one number.")
    elif not any(char.lower() for char in x):
        print("Password should contain at least one lowercase letter.")
    elif not any(char in sp_char for char in x):
        print("Password should contain at least one special character.")
    else:
        print("Strong password.")

check_password()# calling function to check password strength
    
     


```
