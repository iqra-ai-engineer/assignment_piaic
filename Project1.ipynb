**Importing Prompt template and llm chain**


```python
from langchain_core.prompts import PromptTemplate
from langchain.chains import LLMChain
```

**Importing chatgooglegenerativeai class**

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

**Connected Api-Key**

```python
from google.colab import userdata
GOOGLE_API_KEY=userdata.get('Gemini-Api-Key')
```

**Object of langchain**

```python
model : ChatGoogleGenerativeAI = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash-exp",
    api_key=GOOGLE_API_KEY,
    temperature=0.5,
    max_output_tokens=100,
)
```

**Create a prompt template**

```python
prompt_template = PromptTemplate(
    input_variables=["question"],
    template="You are a helpful assistant. Answer the following question:\n\n{question}"
)
```

**Build the LangChain Pipeline**

```python
chain = LLMChain(llm=model, prompt=prompt_template)
```

**Run the First question**

```python
question1 = "What is Langchain?"
response = chain.run({"question": question1})
print("Answer:", response)
```

**Run the second question**

```python
question2 = "What is the use of Langchain?"
response = chain.run({"question": question2})
print("Answer:", response)
```

**Run the third question**

```python
question3 = "What is the difference between Langgraph and Langchain?"
response = chain.run({"question": question3})
print("Answer:", response)
```
